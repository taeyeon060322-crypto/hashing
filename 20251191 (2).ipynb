#7
체이닝에서 새로운 키를 리스트의 맨 앞에 삽입하면
바로 연결만 하면 되므로 삽입 시간이 O(1)이다.

하지만 맨 뒤에 삽입하면 리스트의 끝까지 이동해야 하므로
연결 리스트 길이를 n라 할 때 O(n)의 시간이 필요하다.

따라서 맨 뒤 삽입은 삽입 효율이 감소한다.
#8
체이닝에서 연결 리스트를 정렬된 형태로 유지하면
검색 효율은 좋아지지만 삽입 효율은 나빠진다.

- 삽입:
정렬된 위치를 찾아 삽입해야 하므로 시간이 증가한다.
기존 O(1) 삽입이 O(n)가 될 수 있다.

- 삭제:
삭제할 위치를 찾는 과정이 빨라질 수 있으므로 약간 효율이 향상된다.

- 성공적인 검색:
탐색 중 현재 값이 찾는 값보다 커지면 중단할 수 있으므로
평균 비교 횟수가 감소하여 더 효율적이다.

- 실패하는 검색:
끝까지 탐색하지 않고 중간에 중단 가능하므로
더 효율적이다.
